# K-Means

```python
import numpy as np
```

```python
def kmeans(X, c, max_epochs=100, verbose=False):
    c = c.copy() # don't change the original list
    M = X.shape[0]
    K = c.shape[0]
    for t in range(max_epochs):
        C = [[] for _ in range(K)]
        cost = 0
        for i in range(M):
            mindist = np.infty # infinity
            for k in range(K):
                euclidean_dist = np.sqrt(np.sum((X[i,] - c[k,])**2))
                if euclidean_dist < mindist:
                    best_k = k
                    mindist = euclidean_dist
            cost += mindist**2
            C[best_k].append(i)
        for k in range(K):
            c[k,] = np.mean(X[C[k],], axis=0)
        if verbose:
            print("t =", t, "assignments:\n", C, "\ncluster centres:\n", c)
        if t > 0 and C_prev == C: # why t > 0? Don't check on the first iteration; there is no C_prev
            break # for t
        C_prev = [l.copy() for l in C] # deep-copy the list so it won't get modified
    return t, C, c, cost
```

```python
X = np.array([[2.5, 2.7],
             [2.7,3.1],
             [3.0,3.0],
             [-5.5,2.6],
             [-6.0,3.0],
             [-4.8,2.6],
             [-4.5,2.8],
             [-2.0,-9.3],
             [-1.5,-13.6],
             [-1.7,-7.4]])
c = np.array([X[2], X[3], X[4]])
print("Initial cluster centres:\n", c)
```

```python
kmeans(X, c, verbose=True)
```
